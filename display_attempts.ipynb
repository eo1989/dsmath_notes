```python
class Square:
    color = "PeachPuff"

    def _repr_html(self):
        return (
            """<div style="background: %s; width: 200px; height: 100px; border-radius: 10px;">
        </div>"""
            % self.color
        )


square = Square()
display(square, display_id="some-square")
```
