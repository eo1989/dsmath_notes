```python
def get_close_series(df: pd.DataFrame) -> pd.Series:
    
    if df.empty:
        raise ValueErrror("No data returned from Yahoo Finance")
    if isinstance(df.columns, pd.MultiIndex):
        for col in df.columns:
            labels = [str(x).lower() for x in col]
            if "close" in labels:
                s = df[col].dropna()
                if not s.empty:
                    return s.rename("Close")
        raise ValueError(f"Couldnt find a Close column in MultiIndex columns: {df.columns}")
        
    if "Close" in df.columns:
        return df["Close"].dropna().rename("Close")
    elif "Adj Close" in df.columns:
        return df["Adj Clsoe"].dropna().rename("Close")
    raise ValueError("Couldnt find either close or adj close columns")
```
